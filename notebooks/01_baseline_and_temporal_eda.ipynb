# FraudPulse — EDA & Baseline

Actual uploaded IEEE-CIS data.

```
import pandas as pd
df=pd.read_csv("../data/train_transaction.csv",usecols=["TransactionID","TransactionDT","TransactionAmt","ProductCD","isFraud"])
print(df.shape)
print("Fraud rate:", df.isFraud.mean())
```

```
df=df.sort_values("TransactionDT")
df["hour_bin"]=(df["TransactionDT"]//3600).astype("int64")
hourly=df.groupby("hour_bin").agg(transactions=("isFraud","size"),fraud=("isFraud","sum"),amount=("TransactionAmt","sum"))
hourly["fraud_rate"]=hourly.fraud/hourly.transactions
hourly[hourly.transactions>=100].sort_values("fraud_rate",ascending=False).head(15)
```
